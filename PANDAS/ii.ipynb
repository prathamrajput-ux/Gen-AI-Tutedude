```python
import numpy as np
import pandas as pd
```

```python
#series
state = ['Jharkhand',"Chatisagarh",'Uttar Pradesh']
pd.Series(state)
```

```python
#cahnge dtype
run = [1,2,3,4]
pd.Series(run , dtype = 'int32')
```

```python
#puting index and data
sub = ['dm','aoa','blockchain']
mark = [50,70,90]
pd.Series(index = sub ,data =mark,dtype='int32')
```

```python
#using numpy array
value = np.arange(20,28)
pd.Series(value)  
```

```python
#dictionary using numpy
data = {'suv':'creata','4X4':'Scorpio lifestyle'}
pd.Series(data)
```

```python
value = [1,23 ,4,5]
pd.Series(value,index=['a','b','c','d'])
```

```python
#scalar value
pd.Series(4,index=['a','b','c'])
```
